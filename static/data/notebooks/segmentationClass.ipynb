```python
import os
import pandas as pd
import pickle
from pathlib import Path

BASE_DIR = r"C:\Users\vdela\Desktop\UrbanGazeVis\static\images\images\Disorder"

def inspect_dataframe(df, source_file):
    print(f"\n{'='*60}")
    print(f"FILE: {source_file}")
    print(f"Shape: {df.shape[0]} rows x {df.shape[1]} cols")
    print(f"{'='*60}")
    for col in df.columns:
        dtype = df[col].dtype
        n_unique = df[col].nunique(dropna=False)
        null_count = df[col].isna().sum()
        # Solo muestra unique values si son pocos
        if n_unique <= 30:
            uniques = sorted(df[col].dropna().unique().tolist())
            print(f"  [{dtype}] {col}  →  {n_unique} uniques  ({null_count} nulls)")
            print(f"           {uniques}")
        else:
            sample = df[col].dropna().unique()[:5].tolist()
            print(f"  [{dtype}] {col}  →  {n_unique} uniques  ({null_count} nulls)  sample: {sample}")

# ── CSV ─────────────────────────────────────────────────────────
csv_dir = Path(BASE_DIR) / "csv"
if csv_dir.exists():
    for csv_file in sorted(csv_dir.glob("*.csv")):
        try:
            df = pd.read_csv(csv_file)
            inspect_dataframe(df, csv_file.name)
        except Exception as e:
            print(f"ERROR reading {csv_file.name}: {e}")

# ── PKL ─────────────────────────────────────────────────────────
pkl_dir = Path(BASE_DIR) / "pkl"
if pkl_dir.exists():
    for pkl_file in sorted(pkl_dir.glob("*.pkl")):
        try:
            with open(pkl_file, "rb") as f:
                obj = pickle.load(f)
            if isinstance(obj, pd.DataFrame):
                inspect_dataframe(obj, pkl_file.name)
            elif isinstance(obj, dict):
                print(f"\n{'='*60}")
                print(f"FILE: {pkl_file.name}  [dict con keys: {list(obj.keys())[:10]}]")
                print(f"{'='*60}")
            else:
                print(f"\nFILE: {pkl_file.name}  → tipo: {type(obj)}")
        except Exception as e:
            print(f"ERROR reading {pkl_file.name}: {e}")
```

```python
import pandas as pd
import numpy as np
import os
import joblib

def actualizar_df_grouped_por_imagename():
    # 1. Rutas exactas a tus archivos
    ruta_df = r"C:\Users\vdela\Desktop\UrbanGazeVis\static\data\csv\df_final1.csv"
    carpeta_csv = r"C:\Users\vdela\Desktop\UrbanGazeVis\static\images\images\Disorder\csv"
    carpeta_pkl = r"C:\Users\vdela\Desktop\UrbanGazeVis\static\images\images\Disorder\pkl"

    print("1. Cargando el dataset original...")
    df = pd.read_csv(ruta_df)

    print("2. Inicializando datos Grouped limpios (ADE20K queda intacto)...")
    df["class_id_Disorder"] = "unknown"
    df["main_class_Disorder"] = "unknown"
    df["hex_color_Disorder"] = "#999999"
    df["class_ratio_Disorder"] = 0.0

    FRONTEND_W = 800.0
    FRONTEND_H = 600.0

    # =======================================================
    # LA CLAVE: Iteramos por ImageName, igual que tu script de ADE20K
    # =======================================================
    nombres_unicos = df["ImageName"].dropna().unique()
    print(f"3. Procesando {len(nombres_unicos)} imágenes basadas en ImageName...")

    for i, img_name in enumerate(nombres_unicos):
        
        # Formatear el nombre para que coincida con el archivo (ej. de 3.0 a "3")
        try:
            img_name_val = int(img_name)
        except:
            img_name_val = str(img_name).strip()
            
        archivo_pkl = os.path.join(carpeta_pkl, f"{img_name_val}.pkl")
        archivo_csv = os.path.join(carpeta_csv, f"{img_name_val}.csv")

        if not os.path.exists(archivo_pkl) or not os.path.exists(archivo_csv):
            if i % 10 == 0:
                print(f"   [!] Faltan archivos para ImageName {img_name_val}, saltando...")
            continue

        if i % 10 == 0:
            print(f"   Cruzando datos para ImageName: {img_name_val}.pkl ...")

        try:
            # A. Leer el diccionario (csv)
            df_ratios = pd.read_csv(archivo_csv, sep=';')
            color_map = {}
            for _, row in df_ratios.iterrows():
                try:
                    c_id = int(float(row['class_id']))
                    grupo = str(row['group_name']).strip().lower()
                    ratio_str = str(row['ratio']).replace(',', '.')
                    ratio = float(ratio_str) / 100.0 if float(ratio_str) > 1 else float(ratio_str)
                    hex_val = str(row['hex_color']).strip()
                    color_map[c_id] = {'group_name': grupo, 'ratio': ratio, 'hex': hex_val}
                except: pass
            
            # B. Leer la matriz PKL
            segmentation_array = joblib.load(archivo_pkl)
            img_h, img_w = segmentation_array.shape

            # =======================================================
            # FILTRAMOS POR ImageName (El error estaba aquí antes)
            # =======================================================
            mascara_img = df["ImageName"] == img_name
            indices_gaze = df[mascara_img].index
            
            for idx in indices_gaze:
                if pd.isna(df.loc[idx, "pixelX"]) or pd.isna(df.loc[idx, "pixelY"]):
                    continue
                
                px_orig = float(df.loc[idx, "pixelX"])
                py_orig = float(df.loc[idx, "pixelY"])

                # Validamos que el punto no salga de la pantalla
                if px_orig < 0 or px_orig > FRONTEND_W or py_orig < 0 or py_orig > FRONTEND_H:
                    continue  

                # # =======================================================
                # # ESCALADO EXACTAMENTE IGUAL A ADE20K (Sin invertir ejes)
                # # =======================================================
                # x_pkl = int((px_orig * img_w) / FRONTEND_W)
                # y_pkl = int((py_orig * img_h) / FRONTEND_H)

                # =======================================================
                # ESCALADO CON INVERSIÓN DEL EJE Y (600 - Y)
                # =======================================================
                x_pkl = int((px_orig * img_w) / FRONTEND_W)
                y_pkl = int(((FRONTEND_H - py_orig) * img_h) / FRONTEND_H)

                x_pkl = min(x_pkl, img_w - 1)
                y_pkl = min(y_pkl, img_h - 1)
                
                class_id_pixel = int(segmentation_array[y_pkl, x_pkl])
                
                # Solo inyectamos los datos nuevos en las columnas Grouped
                # if class_id_pixel in color_map:
                #     datos = color_map[class_id_pixel]
                #     df.loc[idx, "class_id_grouped"] = datos['group_name']
                #     df.loc[idx, "main_class_grouped"] = datos['group_name']
                #     df.loc[idx, "class_ratio_grouped"] = datos['ratio']
                #     df.loc[idx, "hex_color_grouped"] = datos['hex']

                if class_id_pixel in color_map:
                    datos = color_map[class_id_pixel]
                    df.loc[idx, "class_id_Disorder"] = datos['group_name']
                    df.loc[idx, "main_class_Disorder"] = datos['group_name']
                    df.loc[idx, "class_ratio_Disorder"] = datos['ratio']
                    df.loc[idx, "hex_color_Disorder"] = datos['hex']

        except Exception as e:
            print(f"Error procesando ImageName {img_name_val}: {e}")

    print("\n4. Guardando los cambios en df_final1.csv...")
    df.to_csv(ruta_df, index=False)
    print("¡ÉXITO!")

if __name__ == "__main__":
    actualizar_df_grouped_por_imagename()
```

```python
import pandas as pd
import os
import glob

# Ruta a tu carpeta de CSVs agrupados
carpeta_csv = r"C:\Users\vdela\Desktop\UrbanGazeVis\static\images\images\Grouped\csv"

# Un 'set' automáticamente ignora los duplicados
todos_los_grupos = set()

# Buscar todos los archivos .csv en esa carpeta
archivos_csv = glob.glob(os.path.join(carpeta_csv, "*.csv"))

print(f"Analizando {len(archivos_csv)} archivos...")

for archivo in archivos_csv:
    try:
        # Leemos cada csv
        df_temp = pd.read_csv(archivo, sep=';')
        
        # Si tiene la columna group_name, extraemos sus valores únicos
        if 'group_name' in df_temp.columns:
            grupos_en_imagen = df_temp['group_name'].dropna().unique()
            todos_los_grupos.update(grupos_en_imagen)
    except Exception as e:
        print(f"Error leyendo {archivo}: {e}")

# Convertimos a una lista ordenada alfabéticamente
lista_final = sorted(list(todos_los_grupos))

print(f"\n¡Listo! Se encontraron {len(lista_final)} grupos únicos en total:")
for grupo in lista_final:
    print(f"- {grupo}")
```
